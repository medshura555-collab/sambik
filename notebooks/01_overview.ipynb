```python
from pathlib import Path
import pandas as pd

# Путь к срезу датасета
data_path = Path("../data/interim/source.parquet")

# Загружаем данные
df = pd.read_parquet(data_path)

# Выводим основные сведения
print("Датасет успешно загружен!")
print(f"Количество строк: {df.shape[0]}")
print(f"Количество столбцов: {df.shape[1]}")

display(df.head())
```

```python
# Типы данных каждого столбца
print("Типы данных:")
display(df.dtypes.to_frame(name="Тип данных"))

# Количество уникальных значений
print("Количество уникальных значений:")
display(df.nunique().to_frame(name="Уникальных значений"))

# Количество и доля пропусков
missing = pd.DataFrame({
    "Пропусков": df.isna().sum(),
    "Доля пропусков, %": (df.isna().mean() * 100).round(2)
})

print("Пропущенные значения:")
display(missing)

# Общая информация о датасете
print("Общая информация:")
df.info()
```
