```python
from pathlib import Path
import time
import gc
import shutil
import pandas as pd

# Пути к файлам
source_file = Path("../data/raw/source.csv")
output_dir = Path(r"E:\laba_results")
output_dir.mkdir(parents=True, exist_ok=True)

# Проверяем исходный файл
if not source_file.exists():
    raise FileNotFoundError(f"Не найден файл: {source_file.resolve()}")

print(f"Исходный файл: {source_file.resolve()}")
print(f"Размер CSV: {source_file.stat().st_size / 1024**3:.2f} ГБ")
print(f"Свободно на диске E: {shutil.disk_usage(r'E:\\').free / 1024**3:.2f} ГБ")

# Настройки обработки
CHUNK_SIZE = 100_000
start_time = time.perf_counter()

# Читаем CSV частями и сохраняем результат в CSV
csv_output = output_dir / "data.csv"
if csv_output.exists():
    csv_output.unlink()

total_rows = 0
first_chunk = True

for chunk in pd.read_csv(source_file, chunksize=CHUNK_SIZE):
    chunk.to_csv(
        csv_output,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )
    total_rows += len(chunk)
    first_chunk = False
    del chunk
    gc.collect()

csv_write_time = time.perf_counter() - start_time

print(f"Обработано строк: {total_rows:,}")
print(f"CSV сохранён: {csv_output}")
print(f"Время записи CSV: {csv_write_time:.2f} сек.")
print(f"Размер результата: {csv_output.stat().st_size / 1024**2:.2f} МБ")
```

```python
import time
from pathlib import Path

import pyarrow as pa
import pyarrow.csv as pacsv
import pyarrow.feather as feather

# Пути
csv_file = Path(
    r"C:\Users\мирон\Desktop\laba po pitonu\data\raw\source.csv"
)
output_dir = Path(r"E:\laba_results\feather_parts")
output_dir.mkdir(parents=True, exist_ok=True)

if not csv_file.exists():
    raise FileNotFoundError(f"CSV не найден: {csv_file}")

# Удаляем результаты предыдущего запуска
for old_file in output_dir.glob("part_*.feather"):
    old_file.unlink()

print("Начинаем потоковое чтение CSV...")
start = time.perf_counter()

# Читаем CSV порциями
reader = pacsv.open_csv(
    csv_file,
    read_options=pacsv.ReadOptions(
        block_size=8 * 1024 * 1024
    )
)

# Берём первые три столбца
columns = reader.schema.names[:3]
print("Выбранные столбцы:", columns)

total_rows = 0
part_number = 0

for batch in reader:
    # Оставляем только первые три столбца
    table = pa.Table.from_batches([batch]).select(columns)

    output_file = output_dir / f"part_{part_number:05d}.feather"
    feather.write_feather(table, output_file, compression="lz4")

    total_rows += table.num_rows
    part_number += 1

    del table, batch

elapsed = time.perf_counter() - start

print("\nГотово!")
print(f"Обработано строк: {total_rows:,}")
print(f"Создано Feather-файлов: {part_number}")
print(f"Папка результатов: {output_dir}")
print(f"Время чтения и записи: {elapsed:.2f} сек.")
print(
    f"Общий размер результатов: "
    f"{sum(f.stat().st_size for f in output_dir.glob('part_*.feather')) / 1024**2:.2f} МБ"
)
```

```python
import time
from pathlib import Path

import pyarrow as pa
import pyarrow.csv as pacsv
import pyarrow.feather as feather

csv_file = Path(
    r"C:\Users\мирон\Desktop\laba po pitonu\data\raw\source.csv"
)
output_dir = Path(r"E:\laba_results\feather_parts")
output_dir.mkdir(parents=True, exist_ok=True)

if not csv_file.exists():
    raise FileNotFoundError(f"CSV не найден: {csv_file}")

# Удаляем старые части, чтобы не смешивать результаты
for f in output_dir.glob("part_*.feather"):
    f.unlink()

# Определяем названия первых трёх столбцов
reader = pacsv.open_csv(
    csv_file,
    read_options=pacsv.ReadOptions(block_size=8 * 1024 * 1024)
)
columns = reader.schema.names[:3]
reader.close()

print("Выбранные столбцы:", columns)

# 1. Измеряем чтение CSV отдельно
read_start = time.perf_counter()
reader = pacsv.open_csv(
    csv_file,
    read_options=pacsv.ReadOptions(block_size=8 * 1024 * 1024)
)

# 2. Измеряем запись Feather отдельно
read_time = 0.0
write_time = 0.0
total_rows = 0
part_number = 0

while True:
    t0 = time.perf_counter()
    try:
        batch = reader.read_next_batch()
    except StopIteration:
        read_time += time.perf_counter() - t0
        break
    read_time += time.perf_counter() - t0

    table = pa.Table.from_batches([batch]).select(columns)

    t1 = time.perf_counter()
    output_file = output_dir / f"part_{part_number:05d}.feather"
    feather.write_feather(table, output_file, compression="lz4")
    write_time += time.perf_counter() - t1

    total_rows += table.num_rows
    part_number += 1

    del table, batch

reader.close()

# 3. Измеряем чтение сохранённых Feather-файлов отдельно
feather_read_start = time.perf_counter()
feather_rows = 0

for f in sorted(output_dir.glob("part_*.feather")):
    table = feather.read_table(f)
    feather_rows += table.num_rows
    del table

feather_read_time = time.perf_counter() - feather_read_start

total_size = sum(
    f.stat().st_size for f in output_dir.glob("part_*.feather")
)
csv_size = csv_file.stat().st_size

print("\n========== РЕЗУЛЬТАТЫ ==========")
print(f"Строк обработано: {total_rows:,}")
print(f"Столбцов сохранено: {len(columns)}")
print(f"Размер Feather: {total_size / (1024**2):.2f} МБ")
print(f"Время чтения CSV: {read_time:.2f} сек.")
print(f"Время записи Feather: {write_time:.2f} сек.")
print(f"Время чтения Feather: {feather_read_time:.2f} сек.")
print(f"Отношение размера Feather к CSV: {total_size / csv_size:.4f}")
print(f"Файлы сохранены в: {output_dir}")
```

```python
import os
import time
import pandas as pd

csv_file = r"E:\laba_results\data.csv"
pickle_dir = r"E:\laba_results\pickle_parts"

os.makedirs(pickle_dir, exist_ok=True)

# Удаляем старые Pickle-части
for name in os.listdir(pickle_dir):
    path = os.path.join(pickle_dir, name)
    if os.path.isfile(path) and name.endswith(".pkl"):
        os.remove(path)

chunk_size = 100_000
rows_total = 0
part_number = 0

# 1. Запись CSV в Pickle-части
csv_read_time = 0.0
pickle_write_time = 0.0

start_total = time.perf_counter()

reader = pd.read_csv(csv_file, chunksize=chunk_size)

while True:
    start_read = time.perf_counter()
    try:
        chunk = next(reader)
    except StopIteration:
        break
    csv_read_time += time.perf_counter() - start_read

    part_path = os.path.join(
        pickle_dir, f"part_{part_number:05d}.pkl"
    )

    start_write = time.perf_counter()
    chunk.to_pickle(part_path)
    pickle_write_time += time.perf_counter() - start_write

    rows_total += len(chunk)
    part_number += 1

conversion_time = time.perf_counter() - start_total

# 2. Считаем общий размер Pickle
pickle_files = [
    os.path.join(pickle_dir, name)
    for name in os.listdir(pickle_dir)
    if name.endswith(".pkl")
]

pickle_size_mb = sum(
    os.path.getsize(path) for path in pickle_files
) / (1024 ** 2)

csv_size_mb = os.path.getsize(csv_file) / (1024 ** 2)

# 3. Чтение всех Pickle-частей
pickle_read_time = 0.0
rows_read = 0

start_read_all = time.perf_counter()

for path in sorted(pickle_files):
    start_part = time.perf_counter()
    chunk = pd.read_pickle(path)
    pickle_read_time += time.perf_counter() - start_part
    rows_read += len(chunk)
    del chunk

pickle_read_total = time.perf_counter() - start_read_all

# 4. Результаты
print("===== РЕЗУЛЬТАТЫ PICKLE =====")
print(f"Строк записано: {rows_total:,}")
print(f"Строк прочитано: {rows_read:,}")
print(f"Количество частей: {len(pickle_files)}")
print(f"Размер CSV: {csv_size_mb:.2f} MB")
print(f"Размер Pickle: {pickle_size_mb:.2f} MB")
print(f"Время чтения CSV при конвертации: {csv_read_time:.2f} сек")
print(f"Время записи Pickle: {pickle_write_time:.2f} сек")
print(f"Время чтения Pickle: {pickle_read_time:.2f} сек")
print(f"Полное время чтения Pickle: {pickle_read_total:.2f} сек")
print(f"Общее время конвертации: {conversion_time:.2f} сек")
print(f"Отношение размера к CSV: {pickle_size_mb / csv_size_mb:.3f}")
```
